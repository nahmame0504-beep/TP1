```python
import string

# Déclaration de la variable text
text = "Bonjour tout le monde ! Le NLP, c'est VRAIMENT passionnant... n'est-ce pas ?"

# 1. Passage en minuscules
text_lower = text.lower()

# 2. Suppression de la ponctuation
text_clean = text_lower.translate(str.maketrans('', '', string.punctuation))

print("Texte original :", text)
print("Texte nettoyé  :", text_clean)
```

```python
# Partie 3 — Tokenisation et Stopwords
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# 1. Tokenisation du texte nettoyé
tokens = word_tokenize(text_clean, language='french')
print("Tokens originaux :", tokens)

# 2. Chargement de la liste des stopwords français
stop_words = set(stopwords.words('french'))

# 3. Ajout d'un mot personnalisé n'apportant pas de sens contextuel
stop_words.add("vraiment")

# 4. Filtrage des mots vides
filtered_tokens = [w for w in tokens if w not in stop_words]

print("\nTokens après filtrage :", filtered_tokens)

# Calcul du pourcentage de suppression
prop_supprimee = (len(tokens) - len(filtered_tokens)) / len(tokens) * 100
print(f"\nProportion supprimée : {prop_supprimee:.2f}%")
```

```python
# Partie 4 — Lemmatisation
# 1. Chargement du modèle spaCy français
nlp = spacy.load("fr_core_news_sm")

# 2. Reconstitution d'un texte à partir des tokens filtrés
text_filtered_str = " ".join(filtered_tokens)

# 3. Traitement avec spaCy
doc = nlp(text_filtered_str)

# 4. Affichage de la comparaison mot brut vs lemme
print(f"{'Mot original':<15} --> {'Lemme'}")
print("-" * 30)
for token in doc:
    print(f"{token.text:<15} --> {token.lemma_}")
```
